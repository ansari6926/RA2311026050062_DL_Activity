# Plant Seedling Species Classification

This notebook is the run entry point for the repository's independently written training pipeline. It trains either a fine-tuned ResNet-34 or a small CNN baseline on the Kaggle Plant Seedlings Classification images.

Download the labeled dataset and place its species folders under `data/plant-seedlings-classification/train/`. The project reports its own metrics after you run the experiment; no prior run's results are presented as yours.

```python
from pathlib import Path
import subprocess
import sys

DATA_DIR = Path('data/plant-seedlings-classification/train')
if not DATA_DIR.exists():
    raise FileNotFoundError(f'Dataset folder not found: {DATA_DIR.resolve()}')

print('Found species folders:', sorted(p.name for p in DATA_DIR.iterdir() if p.is_dir()))
```

## Train the ResNet-34 model

The training script creates a stratified split using seed 42, saves the best checkpoint based on validation macro F1, and evaluates the held-out test split once. Training may take a while on CPU.

```python
subprocess.run([
    sys.executable, 'src/train.py',
    '--data-dir', str(DATA_DIR),
    '--model', 'resnet34',
    '--epochs', '12',
], check=True)
```

## Train the from-scratch baseline

Run separately so its metrics and saved artifacts have their own run folder.

```python
subprocess.run([
    sys.executable, 'src/train.py',
    '--data-dir', str(DATA_DIR),
    '--model', 'small-cnn',
    '--epochs', '20',
], check=True)
```

## Review the results

Each run writes `results.json`, per-class metrics, a confusion matrix, and learning curves in a timestamped folder under `runs/`. Use those generated artifacts to complete `REPORT.md`. In particular, inspect which classes the model confuses and avoid using the test set to tune model choices.
